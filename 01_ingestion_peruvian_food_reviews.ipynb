```python
%sql --name peruvian_food_volume
CREATE VOLUME IF NOT EXISTS workspace.default.peruvian_food_volume;
```

```python
from pyspark.sql import functions as F
```

```python


catalog = "workspace"
schema = "default"
volume = "peruvian_food_volume"

base_path = f"/Volumes/{catalog}/{schema}/{volume}"

print(base_path)
```

```python
restaurantes = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("multiLine", "true")
    .option("escape", '"')
    .load(f"{base_path}/restaurantes.csv")
)
```

```python
reviews = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("multiLine", "true")
    .option("escape", '"')
    .load(f"{base_path}/reviews.csv")
)

display(reviews)
```

```python
model = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("multiLine", "true")
    .option("escape", '"')
    .load(f"{base_path}/model.csv")
)

display(model)
```

```python
restaurantes = spark.read.csv("/Volumes/workspace/default/peruvian_food_volume/restaurants.csv", header=True, inferSchema=True)
print("RESTAURANTES")
print("Filas:", restaurantes.count())
print("Columnas:", len(restaurantes.columns))
print()

reviews = spark.read.csv("/Volumes/workspace/default/peruvian_food_volume/reviews.csv", header=True, inferSchema=True)
print("REVIEWS")
print("Filas:", reviews.count())
print("Columnas:", len(reviews.columns))
print()

model = spark.read.csv("/Volumes/workspace/default/peruvian_food_volume/model.csv", header=True, inferSchema=True)
print("MODEL")
print("Filas:", model.count())
print("Columnas:", len(model.columns))
```

```python
print("=== SCHEMA RESTAURANTES ===")
restaurantes.printSchema()

print("=== SCHEMA REVIEWS ===")
reviews.printSchema()

print("=== SCHEMA MODEL ===")
model.printSchema()
```

```python
print("COLUMNAS RESTAURANTES:")
print(restaurantes.columns)

print("\nCOLUMNAS REVIEWS:")
print(reviews.columns)

print("\nCOLUMNAS MODEL:")
print(model.columns)
```
